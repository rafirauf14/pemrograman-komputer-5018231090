#Task 1: Speed Unit Conversation

```python
speed_kmh = float(input("Enter speed [km/h]:"))

speed_ms = speed_kmh / 3.6

print(f"Speed = {speed_ms:.2f} m/s")
```

#Task 2 : Power Unit Conversion

```python
power_kw = float(input("Enter power [kW]:"))

power_w = power_kw*1000

print(f"Power={power_w:.2f}W")
```

#Task 3 : Electrical Power and Energy Calcuator


```python
voltage= float(input("Enter voltage [V]:"))
current= float(input("Enter current [A]"))
operating_time= float(input("Enter oprating time [hour]"))

power= voltage * current
energy= power * operating_time

print("=====ELECTRICAL CALCULATION=====")
print(f"Voltage={voltage:.2f} V")
print(f"Current={current} A")
print(f"Power={power:.2f} W")
print(f"Operating Time={operating_time:.2f} hour")
print(f"Energy={energy:.2f} Wh")
```

#Task 4 : Comparison and Logical Operators

```python
power_limit= 200
voltage_limit= 24

is_high_power= power>power_limit
is_high_voltage= voltage>voltage_limit
print(f"High power:{is_high_power}")
print(f"High voltage:{is_high_voltage}")
print(f"Both conditions:{is_high_power and is_high_voltage}")
print(f"Any condition:{is_high_power or is_high_voltage}")
```

1. Apa perbedaan int dan float?

"int" digunakan untuk bilangan bulat, sedangkan "float" digunakan untuk bilangan yang memiliki nilai desimal.

Karena input() menghasilkan data bertipe string. Oleh karena itu, data perlu dikonversi menjadi float agar dapat digunakan dalam perhitungan numerik.

3. Apa perbedaan oprator = dan == dalam python?

Operator= digunakan untuk memberikan nilai kepada variabel, sedangkan == digunakan untuk membandingkan apakah dua nilai sama.

4. Apa fungsi operator and dan or?

Operator and digunakan ketika semua kondisi harus bernilai True. Operator or digunakan ketika minimal salah satu kondisi bernilai True.
